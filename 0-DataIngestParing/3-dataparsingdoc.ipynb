### Processing Word Files

```python
from langchain_community.document_loaders import Docx2txtLoader, UnstructuredWordDocumentLoader
```

```python
docx_loader = Docx2txtLoader("data/word_files/proposal.docx")
docx = docx_loader.load()
print(f"Length of docx {len(docx)}")
print(f"Page Content {docx[0].page_content[:100]}")
print(f"Metadata: {docx[0].metadata}")
```

```python
un_docx_loader = UnstructuredWordDocumentLoader("data/word_files/proposal.docx", mode="elements")
un_docx = un_docx_loader.load()
print(f"Length of docx {len(un_docx)}")
for i, doc in enumerate(un_docx):
    print(f"Page Content {i+1}: {un_docx[i].page_content[:100]}")
    print(f"Metadata {i+1}: {un_docx[i].metadata.get('category', 'unknown')}")
```
